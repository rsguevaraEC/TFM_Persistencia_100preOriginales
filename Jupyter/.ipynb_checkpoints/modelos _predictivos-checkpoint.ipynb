```python
import pandas as pd
import psycopg2
from sqlalchemy import create_engine

# Conexión a PostgreSQL
engine = create_engine("postgresql+psycopg2://postgres:ASUS@localhost:5433/postgres")

# Cargar datos normalizados
query = "SELECT marca, fecha, ventas, tipo FROM spo.aeade_ventas_normalizado WHERE tipo='acumulado';"
df = pd.read_sql(query, engine)

# Convertir fecha a datetime y ordenar
df['fecha'] = pd.to_datetime(df['fecha'])
df = df.sort_values(['marca', 'fecha'])

# Visualizar estructura
print(df.head())
```

```python
df_chevrolet = df[df['marca'] == 'CHEVROLET']
```

```python
from statsmodels.tsa.arima.model import ARIMA
```

```python
from prophet import Prophet
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA

# Preparar serie temporal anual
serie = df_chevrolet.set_index('fecha')['ventas']
serie.index = pd.to_datetime(serie.index, format='%Y')  # convertir año a datetime

# Entrenar modelo ARIMA
modelo_arima = ARIMA(serie, order=(1,1,1))
resultado_arima = modelo_arima.fit()

# Predicción a 3 años
pred_arima = resultado_arima.forecast(steps=3)

# Mostrar resumen y resultados
print(resultado_arima.summary())
print("\nPredicciones futuras:")
print(pred_arima)

# Visualizar
plt.figure(figsize=(10,5))
plt.plot(serie, label='Histórico')
plt.plot(pd.date_range(serie.index[-1], periods=3, freq='Y'), pred_arima, label='Predicción ARIMA', color='red')
plt.title('Predicción de ventas Chevrolet (anual)')
plt.legend()
plt.show()

```
