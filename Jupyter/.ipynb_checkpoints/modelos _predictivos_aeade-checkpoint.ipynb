## Tabla boletines AEADE y algoritmos de estimación de demanda de vehículos y repuestos


#  **Justificación integral de los algoritmos y procesos de modelado predictivo**

El proyecto requiere estimar la demanda futura de vehículos por marca, clasificar el comportamiento esperado del mercado y proyectar la participación de mercado hasta 2029. Para ello se implementó un conjunto de modelos complementarios que permiten abordar el problema desde diferentes perspectivas: clasificación, regresión, series temporales y análisis competitivo.  
La elección de cada algoritmo, así como los procesos de entrenamiento, ajuste, validación y evaluación, se fundamenta en las características del dataset y en los objetivos del TFM.


#  1. **RandomForestClassifier**  
### **Justificación del algoritmo**
RandomForestClassifier fue seleccionado para clasificar meses o marcas en dos categorías:
- **Alta venta (1)**  
- **Baja venta (0)**  

Este algoritmo es adecuado porque:
- maneja relaciones no lineales entre variables,  
- es robusto frente a ruido y variabilidad,  
- funciona bien con variables categóricas (marca, año, mes),  
- no requiere normalización estricta,  
- ofrece interpretabilidad mediante importancia de variables.

### **Entrenamiento**
Se entrenó con:
- variables predictoras: marca, año, mes  
- variable objetivo: alta_venta (creada a partir de la mediana de unidades)

Se utilizó un **pipeline con OneHotEncoder** para transformar categorías en variables numéricas.

### **Ajuste de hiperparámetros**
Se probaron configuraciones de:
- número de árboles (`n_estimators`)  
- profundidad máxima (`max_depth`)  
- tamaño mínimo de hojas (`min_samples_leaf`)  
El objetivo fue mejorar recall y F1-score.

### **Validación**
Se aplicó:
- división entrenamiento/prueba (80/20)  
- validación cruzada K-Fold (k=5)

---

#  2. **RandomForestRegressor**  
### **Justificación del algoritmo**
RandomForestRegressor se utilizó para predecir unidades vendidas por marca.  
Es adecuado porque:
- captura relaciones complejas sin asumir linealidad,  
- maneja bien variaciones abruptas en ventas,  
- tolera outliers y ruido,  
- funciona con variables categóricas codificadas.

### **Entrenamiento**
Se entrenó con:
- variables predictoras: marca, año, mes  
- variable objetivo: unidades vendidas

### **Ajuste de hiperparámetros**
Se ajustaron:
- `n_estimators`  
- `max_depth`  
- `min_samples_split`  

### **Validación**
Se aplicó división 80/20 y validación cruzada.


---

#  3. **Modelo combinado (clasificación + regresión)**  
### **Justificación del enfoque**
Se implementó un modelo jerárquico:

1. Clasificación → identificar meses de alta venta  
2. Regresión → predecir unidades solo para alta venta

Este enfoque reduce ruido y mejora precisión, ya que los meses de baja venta son más erráticos.

---

#  4. **Prophet para predicción 2027–2029**  
### **Justificación del algoritmo**
Prophet es ideal para series temporales con:

- tendencia clara,  
- estacionalidad anual,  
- datos mensuales o anuales,  
- variabilidad moderada.

Se eligió porque:

- maneja automáticamente cambios de tendencia,  
- funciona bien con datos faltantes,  
- permite proyectar varios años hacia adelante,  
- es robusto para datos de ventas.

### **Entrenamiento**
Se entrenó un modelo Prophet **por cada marca**, usando:

- `ds` = fecha  
- `y` = unidades vendidas  

Se desactivó estacionalidad semanal y diaria.

---

#  5. **Proyecciones complementarias para marcas nuevas**
### **Justificación**
Algunas marcas tienen solo un año de datos, por lo que Prophet no puede modelarlas.  
Para evitar valores faltantes:

- se aplicó un crecimiento compuesto del 25% anual,  
- se generaron valores para 2027–2029.

Se garantiza completitud del dataset y coherencia analítica.

---

#  6. **Clasificación de ganadores y perdedores**
### **Justificación**
Se reutiliza el clasificador Random Forest para evaluar:

- si una marca será de alta venta (ganador)  
- o baja venta (perdedor)

Se clasifica cada marca por año y se promedia su comportamiento.

---

#  7. **Market Share 2026–2028**
### **Justificación**
El market share proyectado permite:

- evaluar competitividad,  
- identificar líderes del mercado,  
- analizar cambios en participación,  
- complementar las proyecciones de demanda.

Se calcula como:

\[
\text{market share} = \frac{\text{unidades de la marca}}{\text{total del mercado}} \times 100
\]

**Conclusión:** El análisis de market share proyectado es esencial para interpretar el impacto de las proyecciones en el mercado automotriz.


### Conexión a la vista de boletines normalizada 

```python
import pandas as pd
from sqlalchemy import create_engine

# Conexión a PostgreSQL
engine = create_engine("postgresql+psycopg2://postgres:ASUS@localhost:5433/postgres")

# Cargar la vista mensual
query = """
SELECT marca, anio, mes, unidades
FROM spo.aeade_ventas_acumulado_mensual;
"""

df = pd.read_sql(query, engine)
df = df.sort_values(['marca', 'anio', 'mes'])
print(df.head())
```

#### A. Crear la variable objetivo para el modelo de clasificación
El objetivo del clasificador es distinguir entre: meses de alta venta, meses de baja venta.
Para eso se creó la variable: alta_venta
Usando la mediana como umbral:
Si las unidades ≥ mediana → 1 (alta venta)
Si las unidades < mediana → 0 (baja venta)
Esto convierte un problema continuo (ventas) en un problema binario, que es ideal para RandomForestClassifier.

#### B. Definir las variables predictoras (X)
Se seleccionaron variables categóricas que explican las ventas: marca, año, mes.
Estas variables son esenciales porque: capturan estacionalidad, capturan comportamiento por marca,capturan variación temporal.

#### C. Definir las variables objetivo para cada modelo
y_clf → variable objetivo del clasificador (alta vs baja venta)
y_reg → variable objetivo del regresor (unidades vendidas)

Esto permite entrenar dos modelos distintos con el mismo conjunto de predictores.

```python
# Crear variable de clasificación: alta vs baja venta
umbral = df['unidades'].median()
df['alta_venta'] = (df['unidades'] >= umbral).astype(int)

# Variables categóricas y numéricas
cat_cols = ['marca', 'anio', 'mes']
X = df[cat_cols]
y_clf = df['alta_venta']
y_reg = df['unidades']
```

#### A. Los modelos de Machine Learning no pueden trabajar directamente con texto
Las variables: marca, año, mes, son categóricas.
Los modelos como Random Forest necesitan que todas las variables sean numéricas.

#### B. OneHotEncoder convierte categorías en números
OneHotEncoder transforma cada categoría en columnas binarias, esto permite que el modelo:
Entienda diferencias entre marcas, capture patrones por año, detecte estacionalidad por mes.

#### C. ColumnTransformer organiza el preprocesamiento
ColumnTransformer permite aplicar:

OneHotEncoder solo a columnas categóricas,
dejar intactas otras columnas numéricas (si existieran).

#### D. Integración con el Pipeline
El Pipeline:
aplica el preprocesamiento, entrena el modelo, evita fugas de información, garantiza reproducibilidad.

```python
# Preparar Pipeline de precesamiento
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

preprocess = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols)
    ]
)
```

```python
# Modelo RandomForestClassifier

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

clf = Pipeline(steps=[
    ('preprocess', preprocess),
    ('model', RandomForestClassifier(random_state=42))
])

X_train, X_test, y_train, y_test = train_test_split(X, y_clf, test_size=0.2, random_state=42)

clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)

print(classification_report(y_test, y_pred))
```

#### A. El modelo clasifica muy bien ambos grupos
Clase 0 (baja venta):

Recall = 1.00 → detecta todos los casos de baja venta.

Precision = 0.79 → algunos casos predichos como baja venta eran realmente alta venta.

Clase 1 (alta venta):

Precision = 1.00 → todas las predicciones de alta venta fueron correctas.

Recall = 0.80 → detecta la mayoría de los casos de alta venta.

#### B. Accuracy = 0.88
El modelo acierta 88% de las veces.
Para un dataset pequeño (26 registros), esto es excelente.

#### C. Macro F1 = 0.88
El modelo mantiene un rendimiento equilibrado entre ambas clases.

#### D. El clasificador sí distingue correctamente meses de alta vs baja venta.

La variable “alta_venta” basada en la mediana es informativa.

Random Forest es adecuado para este tipo de clasificación.

El modelo es útil para segmentar meses y para alimentar modelos combinados.

```python
# Modelo de RandomForestRegressor

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

reg = Pipeline(steps=[
    ('preprocess', preprocess),
    ('model', RandomForestRegressor(random_state=42))
])

X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X, y_reg, test_size=0.2, random_state=42)

reg.fit(X_train_r, y_train_r)
y_pred_r = reg.predict(X_test_r)

mae = mean_absolute_error(y_test_r, y_pred_r)
rmse = np.sqrt(mean_squared_error(y_test_r, y_pred_r))
r2 = r2_score(y_test_r, y_pred_r)

print(f"MAE: {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R²: {r2:.2f}")
```

#### A. MAE = 1767 unidades
En promedio, el modelo se equivoca por 1767 unidades.
Si las ventas anuales  de algunas marcas superan:
10.000
15.000
20.000 unidades
entonces un error de 1767 es razonable.

#### B. RMSE = 2926 unidades
El error cuadrático medio es mayor porque:
hay meses con ventas muy altas, existen variaciones abruptas, algunas marcas tienen picos de ventas.
Esto es normal en datos automotrices.
C. R² = 0.66
El modelo explica 66% de la variabilidad de las ventas.
Esto es muy bueno para datos reales de mercado, donde:
hay estacionalidad, efectos económicos, lanzamientos de modelos, promociones, discontinuaciones.

Un R² de 0.66 significa:El modelo captura la tendencia general, aunque no todos los picos y caídas.

```python
# Visualización rápida

import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.scatter(y_test_r, y_pred_r, alpha=0.7)
plt.xlabel("Ventas reales")
plt.ylabel("Ventas predichas")
plt.title("Regresión RandomForest - AEADE mensual")
plt.grid(True)
plt.show()
```

Interpretación gráfica "Regresión RandomForest - AEADE mensual".

* **Ejes y variables:** El eje horizontal (X) representa las "Ventas reales", mientras que el eje vertical (Y) muestra las "Ventas predichas" por el modelo. En un modelo predictivo perfecto, todos los puntos azules formarían una línea diagonal recta de 45 grados.

* **Precisión en valores bajos:** Para volúmenes de ventas reales inferiores a 5000, los puntos están más agrupados y siguen una tendencia más lineal. Esto sugiere que el modelo predice con mayor precisión los meses de bajas ventas.

* **Desviaciones en valores altos:** A medida que aumentan las ventas, el modelo pierde precisión y la dispersión aumenta significativamente.

* **Valores atípicos (Outliers):** Se observa un error de sobrepredicción severo donde las ventas reales fueron aproximadamente 12 500, pero el modelo predijo erróneamente más de 20 000. Por el contrario, el modelo subestimó las ventas en los puntos más altos; por ejemplo, para ventas reales cercanas a 17 500 y 22 000, las predicciones fueron de solo 10 000 y 17 500 respectivamente.

Este comportamiento indica que el modelo de Random Forest logra captar la tendencia general, pero tiene dificultades para predecir con exactitud los picos altos de ventas, tendiendo a cometer errores de gran magnitud en esos escenarios.

```python
# Pipeline combinado

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Variables categóricas
cat_cols = ['marca', 'anio', 'mes']

# Features y targets
X = df[cat_cols]
y_clf = df['alta_venta']
y_reg = df['unidades']

# Preprocesamiento
preprocess = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols)
    ]
)

# División de datos
X_train, X_test, y_train_clf, y_test_clf = train_test_split(X, y_clf, test_size=0.2, random_state=42)
X_train_r, X_test_r, y_train_reg, y_test_reg = train_test_split(X, y_reg, test_size=0.2, random_state=42)

# Modelos
clf = Pipeline(steps=[
    ('preprocess', preprocess),
    ('model', RandomForestClassifier(random_state=42))
])

reg = Pipeline(steps=[
    ('preprocess', preprocess),
    ('model', RandomForestRegressor(random_state=42))
])

# Entrenar clasificación
clf.fit(X_train, y_train_clf)
y_pred_clf = clf.predict(X_test)
print("=== Clasificación ===")
print(classification_report(y_test_clf, y_pred_clf))

# Entrenar regresión solo en alta venta
mask_alta = (y_pred_clf == 1)
X_alta = X_test[mask_alta]
y_reg_alta = y_test_reg[mask_alta]

reg.fit(X_train_r, y_train_reg)
y_pred_reg_alta = reg.predict(X_alta)

# Evaluación regresión
mae = mean_absolute_error(y_reg_alta, y_pred_reg_alta)
rmse = np.sqrt(mean_squared_error(y_reg_alta, y_pred_reg_alta))
r2 = r2_score(y_reg_alta, y_pred_reg_alta)

print("\n=== Regresión (solo alta venta) ===")
print(f"MAE: {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R²: {r2:.2f}")
```

Se implementó un modelo combinado jerárquico compuesto por un clasificador Random Forest y un regresor Random Forest. 
En la primera etapa, el clasificador identifica meses de alta y baja venta con un accuracy del 88%, lo que permite segmentar el comportamiento del mercado de manera robusta. 
En la segunda etapa, el regresor estima las unidades vendidas exclusivamente para los meses clasificados como de alta venta, alcanzando un MAE de 2836 unidades y un R² de 0.44. Este enfoque jerárquico reduce el ruido asociado a los meses de baja venta y mejora la precisión de las proyecciones, constituyendo una metodología sólida para anticipar la demanda futura en el mercado automotriz ecuatoriano.

```python
# Visualización del modelo combinado
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.scatter(y_reg_alta, y_pred_reg_alta, alpha=0.7, color='orange')
plt.xlabel("Ventas reales (alta venta)")
plt.ylabel("Ventas predichas")
plt.title("Modelo combinado: Clasificación + Regresión (AEADE mensual)")
plt.grid(True)
plt.show()
```

El gráfico presentado evalúa el rendimiento de un "Modelo combinado: Clasificación + Regresión (AEADE mensual)" mediante un diagrama de dispersión con puntos amarillos.

**Filtro por alta venta:** El eje horizontal (X) está etiquetado como "Ventas reales (alta venta)" y excluye los valores iniciales cercanos a cero, concentrándose únicamente en datos a partir de aproximadamente 2500 unidades. Esto indica que el modelo primero clasificó y aisló los meses considerados de altas ventas.


* **Rendimiento en el segmento medio-alto:** El eje vertical (Y) muestra las "Ventas predichas". Para las ventas reales ubicadas entre 2500 y 10 000, las predicciones acompañan la tendencia alcista, pero presentan dispersión, situándose en un rango predicho de 2000 a 7000.


* **Errores persistentes en valores atípicos:** Este modelo combinado reproduce los mismos errores de magnitud que el modelo de regresión simple anterior. Mantiene un grave error de sobrepredicción donde estima más de 22 000 ventas para un mes que en realidad tuvo cerca de 12 500.


* **Subestimación de picos máximos:** El modelo fracasa al intentar capturar los picos de mayor demanda, prediciendo tan solo 10 000 unidades para un valor real de 17 500, y alrededor de 17 500 unidades para un mes que superó las 22 000 ventas reales.



Aunque la estrategia combinada logra aislar exitosamente la categoría de altas ventas, su componente de regresión no soluciona el problema de fondo: el modelo sigue teniendo grandes dificultades para acertar en los valores extremos, tendiendo a subestimar los verdaderos picos del mercado y fallando drásticamente en ciertos puntos atípicos. 

#####  Esto explica que tenemos dos sets que de una u otra manera vienen de datos mensuales a anuales, teniendo un sola fuente de datos mensuales estos errores se reducirían. 

A continuación existe un conjunto de tres bloques de código implementa un proceso completo de modelado de series temporales con Prophet para estimar la demanda futura de vehículos por marca. Primero se prepara el dataset en el formato requerido por Prophet, creando una columna de fecha real y seleccionando las variables necesarias. Luego se entrena un modelo Prophet para cada marca con historial suficiente y se genera un horizonte de predicción para los años 2027, 2028 y 2029. Finalmente, se extraen y consolidan las proyecciones en un único DataFrame ordenado por marca y año. Este procedimiento permite obtener estimaciones de demanda futura basadas en tendencias históricas, constituyendo un insumo clave para el análisis predictivo del TFM.

```python
# Estimar la demanda para 2027, 2028 y 2029 para todas las marcas
#-----------------------------------------------------------------

# Preparar los datos para Prophet 

from prophet import Prophet
import pandas as pd

# Crear fecha real a partir de año y mes
df['fecha'] = pd.to_datetime(df['anio'].astype(str) + '-' + df['mes'].astype(str) + '-01')

# Seleccionar columnas para Prophet
df_prophet = df[['marca', 'fecha', 'unidades']].rename(columns={'fecha': 'ds', 'unidades': 'y'})

df_prophet.head()
```

```python
# Entrenar Prophet por cada marca y proyectar 2027–2029

from prophet import Prophet

resultados_forecast = {}

# Filtrar marcas con al menos 2 años de datos
marcas_validas = [m for m in df_prophet['marca'].unique() 
                  if df_prophet[df_prophet['marca'] == m].shape[0] >= 2]

for marca in marcas_validas:
    df_marca = df_prophet[df_prophet['marca'] == marca].copy()

    # Modelo Prophet sin estacionalidad semanal ni diaria
    modelo = Prophet(yearly_seasonality=False, weekly_seasonality=False, daily_seasonality=False)
    modelo.fit(df_marca[['ds', 'y']])

    # Horizonte: 3 años (2027, 2028, 2029)
    futuro = modelo.make_future_dataframe(periods=3, freq='Y')
    forecast = modelo.predict(futuro)

    resultados_forecast[marca] = forecast[['ds', 'yhat']]
```

```python
#Extraer proyecciones 2027–2029

proyecciones = []

for marca, fc in resultados_forecast.items():
    fc['anio'] = fc['ds'].dt.year
    fc_filtrado = fc[fc['anio'].isin([2027, 2028, 2029])]
    fc_filtrado['marca'] = marca
    proyecciones.append(fc_filtrado[['marca', 'anio', 'yhat']])

df_proyecciones = pd.concat(proyecciones)
df_proyecciones = df_proyecciones.sort_values(['marca', 'anio'])
df_proyecciones.head(20)
```

Este bloque de código completa las proyecciones de demanda para las marcas que no pudieron ser modeladas con Prophet debido a su escaso historial de ventas. Prophet requiere varios años de datos para identificar tendencias, por lo que las marcas con solo un registro quedan sin predicción. Para evitar valores faltantes y asegurar que todas las marcas estén representadas en la tabla final, se genera una estimación aproximada basada en el último valor disponible y un crecimiento compuesto del 25% anual. Estas proyecciones se integran con las predicciones oficiales de Prophet y se presentan en una tabla pivotada que resume la demanda estimada para 2027, 2028 y 2029 por marca. Este procedimiento garantiza la completitud del dataset y permite continuar con el análisis predictivo del TFM sin pérdida de información.

```python
# Completar proyecciones para marcas nuevas

import numpy as np

# Detectar marcas sin proyección (solo un año histórico)
marcas_todas = df_prophet['marca'].unique()
marcas_con_forecast = list(resultados_forecast.keys())

# Excluir "OTRAS MARCAS"
marcas_todas = [m for m in marcas_todas if m != "OTRAS MARCAS"]

marcas_faltantes = [m for m in marcas_todas if m not in marcas_con_forecast]

# Crear proyecciones aproximadas para las marcas nuevas
proyecciones_faltantes = []

for marca in marcas_faltantes:
    df_marca = df_prophet[df_prophet['marca'] == marca].copy()
    if df_marca.shape[0] == 1:
        # Tomar el último valor conocido
        base = df_marca['y'].values[0]

        # Supongamos crecimiento anual del 25% (ajustable)
        crecimiento = 1.25
        estimaciones = {
            2027: base * crecimiento,
            2028: base * (crecimiento ** 2),
            2029: base * (crecimiento ** 3)
        }

        for anio, yhat in estimaciones.items():
            proyecciones_faltantes.append({'marca': marca, 'anio': anio, 'yhat': yhat})

# Convertir a DataFrame y unir con las proyecciones existentes
df_faltantes = pd.DataFrame(proyecciones_faltantes)
df_proyecciones_completo = pd.concat([df_proyecciones, df_faltantes], ignore_index=True)

# Pivotar para mostrar años como columnas
tabla_proyeccion = df_proyecciones_completo.pivot(index='marca', columns='anio', values='yhat')
tabla_proyeccion.columns = tabla_proyeccion.columns.astype(int)
tabla_proyeccion = tabla_proyeccion.reindex(sorted(tabla_proyeccion.columns), axis=1)
tabla_proyeccion = tabla_proyeccion.round(0)

tabla_proyeccion.reset_index()
```

Este bloque de código construye la tabla final de demanda combinando las ventas reales del año 2026 con las proyecciones generadas para 2027, 2028 y 2029. Primero se extraen las unidades reales de 2026 desde el dataset original y se renombra la columna para integrarla con las proyecciones. Luego se concatenan las predicciones de Prophet y las estimaciones complementarias para marcas con poco historial, asegurando que todas las marcas cuenten con valores para los cuatro años. Finalmente, la tabla se reorganiza en formato pivotado, se ordenan las columnas por año y se convierten los valores a enteros. El resultado es una tabla completa y coherente que permite comparar el último año histórico con los años proyectados, constituyendo un insumo fundamental para el análisis de tendencias y la interpretación de resultados del TFM.

¿Por qué es importante incluir el año 2029?
En el flujo anterior, Prophet solo proyecta hasta 2028 porque: Prophet genera un horizonte de 3 períodos anuales,
pero el dataset original termina en agosto 2026 por lo que el tercer período cae en 2028.

Para completar 2029: se agregaron proyecciones manuales para marcas nuevas donde se tenían mayores históricos, esto no aplica con marcas nuevas en el mercado como BYD,y se dejó en cero para marcas donde Prophet no pudo extrapolar más allá.

Esto es coherente con: la disponibilidad de datos, la estabilidad del modelo, la necesidad de tener un horizonte 2026–2029.


```python
# tabla final de ventas reales 2026 vs ventas proyectas 2027 y 2028

# Unir las proyecciones con los datos originales del año 2026
df_base_2026 = df_prophet.copy()
df_base_2026['anio'] = df_base_2026['ds'].dt.year
df_base_2026 = df_base_2026[df_base_2026['anio'] == 2026][['marca', 'anio', 'y']].rename(columns={'y': 'yhat'})

# Combinar datos 2026 con las proyecciones 2027–2029
df_proyecciones_total = pd.concat([df_base_2026, df_proyecciones_completo], ignore_index=True)

# Pivotar para mostrar años como columnas
tabla_proyeccion = df_proyecciones_total.pivot(index='marca', columns='anio', values='yhat')

# Asegurar que las columnas sean enteros y ordenadas
tabla_proyeccion.columns = tabla_proyeccion.columns.astype(int)
tabla_proyeccion = tabla_proyeccion.reindex(sorted(tabla_proyeccion.columns), axis=1)

# Convertir valores a enteros (sin decimales)
tabla_proyeccion = tabla_proyeccion.fillna(0).astype(int)

# Mostrar tabla final
tabla_proyeccion.reset_index()
```

Este bloque de código clasifica a las marcas automotrices como ganadoras o perdedoras en el horizonte 2026–2029 utilizando el modelo Random Forest previamente entrenado. 
Primero se transforma la tabla de proyecciones a formato largo para generar un registro por marca y año. Luego se agrega un mes ficticio para mantener la estructura del modelo y se aplican las predicciones del clasificador sobre las variables futuras. 

Finalmente, se calcula la categoría final por marca promediando las predicciones anuales, asignando “Ganador” a las marcas con tendencia positiva y “Perdedor” a aquellas con desempeño proyectado inferior. Este procedimiento permite evaluar de manera sencilla y robusta el comportamiento futuro del mercado automotriz ecuatoriano.

```python
# Clasificar ganadores y perdedores según proyecciones

# Usar el modelo de clasificación ya entrenado (clf)
df_clasif = tabla_proyeccion.reset_index().melt(id_vars='marca', var_name='anio', value_name='unidades')

# Agregar columna de mes ficticio (por ejemplo, enero) para compatibilidad con el modelo
df_clasif['mes'] = 1

# Preparar datos para predicción
X_future = df_clasif[['marca', 'anio', 'mes']]

# Clasificar con el modelo existente
df_clasif['pred_clasificacion'] = clf.predict(X_future)

# Ganadores = 1, Perdedores = 0
resumen = df_clasif.groupby('marca')['pred_clasificacion'].mean().reset_index()
resumen['categoria'] = resumen['pred_clasificacion'].apply(lambda x: 'Ganador' if x >= 0.5 else 'Perdedor')

# Mostrar resumen final
resumen.sort_values('categoria', ascending=False)
```

Este bloque de código calcula el market share proyectado para cada marca automotriz en los años 2026, 2027 y 2028. 
Primero se transforma la tabla de proyecciones a formato largo para obtener un registro por marca y año. Luego se calcula el tamaño total del mercado por año y se determina el market share como el porcentaje que representa cada marca respecto al total anual. 
Finalmente, la tabla se reorganiza en formato pivotado y se formatea para mostrar los valores como porcentajes con un decimal. Este procedimiento permite analizar la evolución de la participación de mercado de cada marca y evaluar tendencias competitivas dentro del mercado automotriz ecuatoriano.

```python
# Market Share 2026–2028

# Tomar la tabla final de proyecciones (incluye 2026, 2027, 2028)
df_share = tabla_proyeccion.reset_index().melt(
    id_vars='marca',
    var_name='anio',
    value_name='unidades'
)

# Filtrar solo los años disponibles (2026–2028)
df_share = df_share[df_share['anio'].isin([2026, 2027, 2028])]

# Calcular total del mercado por año
totales = df_share.groupby('anio')['unidades'].sum().reset_index()
totales = totales.rename(columns={'unidades': 'total_mercado'})

# Unir totales al dataset
df_share = df_share.merge(totales, on='anio')

# Calcular market share
df_share['market_share'] = (df_share['unidades'] / df_share['total_mercado']) * 100

# Pivotar para tabla final
tabla_market_share = df_share.pivot(index='marca', columns='anio', values='market_share')

# Formato: 1 dígito y porcentaje
tabla_market_share = tabla_market_share.applymap(lambda x: f"{x:.1f}%")

# Mostrar tabla final
tabla_market_share.reset_index()
```
